```python
# ============================================
# Sales Analytics Dashboard
# Phase 2 : Data Cleaning
# ============================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully ✅")
```

```python
df = pd.read_csv("../data/raw/superstore_sales.csv")

print("Dataset Loaded Successfully ✅")
```

```python
sales_df = df.copy()
```

```python
sales_df.isnull().sum()
missing = sales_df.isnull().sum()

missing_percent = (
    missing / len(sales_df) * 100
).round(2)

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percent
})

missing_df
```

```python
duplicates = sales_df.duplicated().sum()

print("Duplicate Rows :", duplicates)

sales_df.drop_duplicates(inplace=True)
```

```python
sales_df.dtypes
```

```python
sales_df["Order Date"] = pd.to_datetime(
    sales_df["Order Date"]
)

sales_df["Ship Date"] = pd.to_datetime(
    sales_df["Ship Date"]
)
sales_df.dtypes
```

```python
numeric_cols = [
    "Sales",
    "Profit",
    "Quantity",
    "Discount"
]

sales_df[numeric_cols].describe()
```

```python
sales_df[sales_df["Sales"] < 0]
sales_df[sales_df["Quantity"] <= 0]
sales_df[sales_df["Discount"] > 1]
```

```python
Q1 = sales_df["Sales"].quantile(0.25)
Q3 = sales_df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = sales_df[
    (sales_df["Sales"] < lower) |
    (sales_df["Sales"] > upper)
]

print(outliers.shape)
```

```python
sales_df["Order Year"] = (
    sales_df["Order Date"].dt.year
)
sales_df["Order Month"] = (
    sales_df["Order Date"].dt.month_name()
)
sales_df["Quarter"] = (
    sales_df["Order Date"].dt.quarter
)
sales_df["Delivery Days"] = (
    sales_df["Ship Date"] -
    sales_df["Order Date"]
).dt.days
sales_df["Profit Margin"] = (
    sales_df["Profit"] /
    sales_df["Sales"]
) * 100
```

```python
sales_df.head()
```

```python
sales_df.to_csv(
    "../data/cleaned/clean_sales_data.csv",
    index=False
)

print("Clean dataset saved successfully! ✅")
```
